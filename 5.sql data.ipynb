```python
import pandas as pd
import pymysql 

# ---------------------------------------------------------
# 1. CONNECT TO MYSQL
# ---------------------------------------------------------

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="your_password",
    database="cart2insight",
    autocommit=True
)

cursor = connection.cursor()
print("Connected to MySQL!")

# ---------------------------------------------------------
# 2. LOAD FEATURE ENGINEERED DATA
# ---------------------------------------------------------

orders = pd.read_csv("data/processed/orders_fe.csv")
customers = pd.read_csv("data/processed/customers_fe.csv")
sellers = pd.read_csv("data/processed/sellers_fe.csv")

print("Loaded feature-engineered datasets!")

# ---------------------------------------------------------
# 3. CREATE TABLES (BEGINNER-FRIENDLY)
# ---------------------------------------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders_fe (
    order_id VARCHAR(50),
    customer_id VARCHAR(50),
    order_status VARCHAR(50),
    order_purchase_timestamp DATETIME,
    order_delivered_customer_date DATETIME,
    order_estimated_delivery_date DATETIME,
    order_value FLOAT,
    delivery_days INT,
    delivery_delay_days INT
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS customers_fe (
    customer_id VARCHAR(50),
    customer_unique_id VARCHAR(50),
    customer_zip_code_prefix INT,
    customer_city VARCHAR(100),
    customer_state VARCHAR(10),
    customer_total_spent FLOAT,
    customer_spending_tier VARCHAR(20)
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS sellers_fe (
    seller_id VARCHAR(50),
    seller_zip_code_prefix INT,
    seller_city VARCHAR(100),
    seller_state VARCHAR(10),
    seller_revenue FLOAT,
    seller_order_count INT,
    seller_avg_review_score FLOAT
);
""")

print("MySQL tables created!")

# ---------------------------------------------------------
# 4. INSERT DATA INTO MYSQL
# ---------------------------------------------------------

def insert_data(df, table_name):
    cols = ",".join(df.columns)
    placeholders = ",".join(["%s"] * len(df.columns))
    query = f"INSERT INTO {table_name} ({cols}) VALUES ({placeholders})"

    for _, row in df.iterrows():
        cursor.execute(query, tuple(row))

    print(f"Inserted {len(df)} rows into {table_name}!")

insert_data(orders, "orders_fe")
insert_data(customers, "customers_fe")
insert_data(sellers, "sellers_fe")

print("All data inserted into MySQL!")

# ---------------------------------------------------------
# 5. CLOSE CONNECTION
# ---------------------------------------------------------

cursor.close()
connection.close()
print("MySQL connection closed!")
```
