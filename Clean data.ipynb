### Cleaning Data ####

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# ---------------------------------------------------------
# 1. LOAD RAW DATA (Correct file → correct variable)
# ---------------------------------------------------------

orders = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_orders_dataset.csv")
order_items = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_order_items_dataset.csv")
order_payments = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_order_payments_dataset.csv")
order_reviews = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_order_reviews_dataset.csv")
customers = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_customers_dataset.csv")
sellers = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_sellers_dataset.csv")
products = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_products_dataset.csv")
product_category_name_translation = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/product_category_name_translation.csv")
geolocation = pd.read_csv("C:/Users/Sakthi/OneDrive/Documents/Bharathi - Guvi-AIML/project final 1/Cart2Insight/Raw data/olist_geolocation_dataset.csv")

print("Loaded raw data for cleaning!")

# ---------------------------------------------------------
# 2. DATASET SUMMARY LOOP (Correct variables)
# ---------------------------------------------------------

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "customers": customers,
    "sellers": sellers,
    "products": products,
    "product_category": product_category_name_translation,
    "geolocation": geolocation
}

for name, data in datasets.items():
    print(f"\n==================== DATASET: {name.upper()} ====================")
    print(f"Shape: {data.shape}")
    print(f"Duplicates: {data.duplicated().sum()}")
    print("Missing Values Summary:")
    null_counts = data.isnull().sum()
    null_pct = (null_counts / len(data)) * 100
    missing_table = pd.DataFrame({
        'Missing Count': null_counts,
        'Missing Pct (%)': null_pct.round(2)
    })
    print(missing_table[missing_table['Missing Count'] > 0])

# ---------------------------------------------------------
# 3. DATE CONVERSIONS (Correct datasets)
# ---------------------------------------------------------

# Orders date columns
date_cols_orders = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols_orders:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

# Order items date column
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"], errors="coerce"
)

# Order reviews date column
order_reviews["review_creation_date"] = pd.to_datetime(
    order_reviews["review_creation_date"], errors="coerce"
)

# ---------------------------------------------------------
# 4. REMOVE DUPLICATES (Correct variables)
# ---------------------------------------------------------

orders = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
order_payments = order_payments.drop_duplicates()
order_reviews = order_reviews.drop_duplicates()
customers = customers.drop_duplicates()
sellers = sellers.drop_duplicates()
products = products.drop_duplicates()
product_category_name_translation = product_category_name_translation.drop_duplicates()
geolocation = geolocation.drop_duplicates()

print("Duplicates removed!")

# ---------------------------------------------------------
# 5. OUTLIER REMOVAL (Correct columns)
# ---------------------------------------------------------

order_items = order_items[order_items["price"] >= 0]
order_items = order_items[order_items["freight_value"] >= 0]

print("Outliers removed!")

# ---------------------------------------------------------
# 6. KNN IMPUTATION (Correct columns)
# ---------------------------------------------------------

imputer = KNNImputer(n_neighbors=3)
order_items[["price", "freight_value"]] = imputer.fit_transform(
    order_items[["price", "freight_value"]]
)

print("KNN imputation completed!")

# ---------------------------------------------------------
# 7. SAVE CLEANED DATA (Correct variables)
# ---------------------------------------------------------

orders.to_csv("data/processed/orders_clean.csv", index=False)
order_items.to_csv("data/processed/order_items_clean.csv", index=False)
order_payments.to_csv("data/processed/order_payments_clean.csv", index=False)
order_reviews.to_csv("data/processed/order_reviews_clean.csv", index=False)
customers.to_csv("data/processed/customers_clean.csv", index=False)
sellers.to_csv("data/processed/sellers_clean.csv", index=False)
products.to_csv("data/processed/products_clean.csv", index=False)
product_category_name_translation.to_csv("data/processed/product_category_name_translation_clean.csv", index=False)
geolocation.to_csv("data/processed/geolocation_clean.csv", index=False)

print("All cleaned datasets saved!")
```
